```python
n = int(input("Enter the number: "))

a, b = 0, 1

print("Fibonacci Sequence:")

while a <= n:
    print(a, end=" ")
    a, b = b, a + b
```

```python
def rotate_list(lst, k):
    k = k % len(lst)
    return lst[-k:] + lst[:-k]

numbers = [1, 2, 3, 4, 5]

k = int(input("Enter rotation positions: "))

result = rotate_list(numbers, k)

print("Rotated List:", result)
```
